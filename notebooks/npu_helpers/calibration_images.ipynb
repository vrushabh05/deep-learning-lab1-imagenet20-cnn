```python
import os
import random
import shutil
from pathlib import Path

train_root = "/content/imagenet_train20a"
calib_dir = "/content/calibration_images"

os.makedirs(calib_dir, exist_ok=True)

all_images = []
for class_dir in Path(train_root).iterdir():
    if class_dir.is_dir():
        for img in class_dir.glob("*"):
            all_images.append(str(img))

print("Total training images:", len(all_images))

random.seed(42)
selected = random.sample(all_images, 200)

for i, img_path in enumerate(selected):
    shutil.copy(img_path, os.path.join(calib_dir, f"calib_{i}.jpg"))

print("Calibration images created:", len(os.listdir(calib_dir)))
```

```python
import zipfile

drive_output = "/content/drive/MyDrive/DATA 255 Lab/outputs_lab1"
os.makedirs(drive_output, exist_ok=True)

zip_path = os.path.join(drive_output, "calibration_images.zip")

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for f in Path(calib_dir).glob("*"):
        z.write(f, arcname=f.name)

print("Calibration zip saved to:", zip_path)
```
